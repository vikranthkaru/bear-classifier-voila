```python
from fastai.vision.all import *
from fastai.vision.widgets import *
import ipywidgets as widgets

path = Path()
learn_inf = load_learner(path / "export.pkl")

btn_upload = widgets.FileUpload()
btn_run = widgets.Button(description="Classify")
out_pl = widgets.Output()
lbl_pred = widgets.Label()


def _extract_upload_bytes(uploader):
    # ipywidgets v8+: value is a tuple/list of dicts or Bunch-like items with 'content'
    val = getattr(uploader, "value", None)
    if isinstance(val, (tuple, list)) and len(val) > 0:
        last = val[-1]
        if isinstance(last, dict) and "content" in last:
            return bytes(last["content"])
        if hasattr(last, "content"):
            return bytes(last.content)
    # ipywidgets v7 fallback: data is a list of byte-like objects
    data = getattr(uploader, "data", None)
    if data:
        return bytes(data[-1])
    return None


def on_click_classify(change):
    raw = _extract_upload_bytes(btn_upload)
    if not raw:
        lbl_pred.value = "Please upload an image first."
        return
    try:
        img = PILImage.create(raw)
        out_pl.clear_output()
        with out_pl:
            display(img.to_thumb(128, 128))
        pred, pred_idx, probs = learn_inf.predict(img)
        lbl_pred.value = f"Prediction: {pred}; Probability: {probs[pred_idx]:.04f}"
    except Exception as e:
        lbl_pred.value = f"Error: {type(e).__name__}: {e}"


btn_run.on_click(on_click_classify)

widgets.VBox(
    [
        widgets.Label("Select your bear!"),
        btn_upload,
        btn_run,
        out_pl,
        lbl_pred,
    ]
)
```
